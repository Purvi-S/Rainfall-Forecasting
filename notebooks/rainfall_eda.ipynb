# Rainfall Forecasting — Exploratory Data Analysis

This notebook explores the weather dataset and walks through the modelling
pipeline (Random Forest, SVM, XGBoost) used to predict whether it will rain
tomorrow.

> Run this from the project root, or the relative paths below won't resolve.

````python
import os, sys
# make the src package importable when running from notebooks/
sys.path.append(os.path.abspath('..'))

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.data_loader import load_data

sns.set_theme(style='whitegrid')
````

## 1. Load the data

If `data/weather.csv` is missing, a synthetic dataset is generated
automatically. Drop in the real Kaggle `weatherAUS.csv` (renamed to
`data/weather.csv`) to analyse the real-world data instead.

````python
df = load_data('../data/weather.csv')
print(df.shape)
df.head()
````

````python
df.describe(include='all').T
````

## 2. Target distribution

Rainy days are the minority class — worth remembering when reading accuracy.

````python
ax = df['RainTomorrow'].value_counts().plot(kind='bar', color=['#4a7db5', '#b5674a'])
ax.set_title('RainTomorrow class balance')
ax.set_ylabel('count')
plt.show()
print((df['RainTomorrow'] == 'Yes').mean())
````

## 3. Correlations between numeric features

````python
num = df.select_dtypes('number')
plt.figure(figsize=(9, 7))
sns.heatmap(num.corr(), annot=False, cmap='coolwarm', center=0)
plt.title('Numeric feature correlations')
plt.show()
````

## 4. Humidity vs. rain tomorrow

Afternoon humidity is one of the strongest single predictors.

````python
sns.boxplot(data=df, x='RainTomorrow', y='Humidity3pm',
            palette=['#4a7db5', '#b5674a'])
plt.title('3pm humidity by next-day rain')
plt.show()
````
